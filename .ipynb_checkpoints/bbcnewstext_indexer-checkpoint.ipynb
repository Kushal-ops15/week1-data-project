```python
import os
import re
from collections import Counter
import pandas as pd

# Path to your text files|
folder_path = 'C:/Users/Kushal  sri sai/week1-data-project/bbc_files'
data = []

# List files in the folder and grab up to 10

files = []

for f in os.listdir(folder_path):
    if f.endswith('.txt'):
        files.append(f)
    if len(files) == 10:
        break

# Simple list of common words to ignore when finding keywords
stop_words = {'the', 'and', 'a', 'of', 'to', 'in', 'is', 'that', 'it', 'for', 'on', 'with', 'as', 'was', 'at', 'by', 'an', 'be', 'are', 'from', 's'}
for filename in files:
    file_path = os.path.join(folder_path, filename)
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as file:
        text = file.read()
        
        # Calculate total word count
        words = re.findall(r'\b\w+\b', text.lower())
        word_count = len(words)
        
        # Filter out numbers and common stop words to find meaningful keywords
      
        filtered_words = []
        
       
        for w in words:
           if w not in stop_words and not w.isdigit():
              filtered_words.append(w)
        top_5 = [word for word, count in Counter(filtered_words).most_common(5)]
        
        data.append({'filename': filename,'word_count': word_count,'top_5_keywords': top_5 })



# Create the structured Pandas DataFrame
df = pd.DataFrame(data)
display(df)
```
